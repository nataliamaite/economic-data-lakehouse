```python

import requests
import pandas as pd

url_selic = (
    "https://api.bcb.gov.br/dados/serie/bcdata.sgs.432/dados"
    "?formato=json&dataInicial=01/01/2025&dataFinal=31/12/2025"
)

response = requests.get(url_selic, timeout=30)
response.raise_for_status()

dados_selic = response.json()

df_selic = pd.DataFrame(dados_selic)

display(df_selic.head())
```

```python

# 1. Criar o schema do projeto
spark.sql("""
    CREATE SCHEMA IF NOT EXISTS workspace.economic_data
""")

# 2. Converter os dados do Pandas para Spark
df_selic_spark = spark.createDataFrame(df_selic)

# 3. Salvar os dados como tabela Delta
(
    df_selic_spark.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable("workspace.economic_data.bronze_selic")
)

print("Tabela Bronze da Selic criada com sucesso!")
```

```python

url_dolar = (
    "https://api.bcb.gov.br/dados/serie/bcdata.sgs.1/dados"
    "?formato=json&dataInicial=01/01/2025&dataFinal=31/12/2025"
)

response_dolar = requests.get(url_dolar, timeout=30)
response_dolar.raise_for_status()

dados_dolar = response_dolar.json()
df_dolar = pd.DataFrame(dados_dolar)

display(df_dolar.head())
```

```python

df_dolar_spark = spark.createDataFrame(df_dolar)

(
    df_dolar_spark.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable("workspace.economic_data.bronze_dolar")
)

print("Tabela Bronze do dólar criada com sucesso!")
```
