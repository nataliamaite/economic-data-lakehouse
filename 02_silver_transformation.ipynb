```python

df_selic_bronze = spark.table(
    "workspace.economic_data.bronze_selic"
)

df_dolar_bronze = spark.table(
    "workspace.economic_data.bronze_dolar"
)

display(df_selic_bronze.limit(5))
display(df_dolar_bronze.limit(5))
```

```python

print("Tipos das colunas da Selic:")
df_selic_bronze.printSchema()

print("Tipos das colunas do dólar:")
df_dolar_bronze.printSchema()
```

```python

from pyspark.sql.functions import col, to_date

df_selic_silver = (
    df_selic_bronze
    .withColumn(
        "data",
        to_date(col("data"), "dd/MM/yyyy")
    )
    .withColumn(
        "valor",
        col("valor").cast("double")
    )
    .dropDuplicates(["data"])
    .filter(
        col("data").isNotNull()
        & col("valor").isNotNull()
    )
)

display(df_selic_silver.limit(10))
```

```python

df_dolar_silver = (
    df_dolar_bronze
    .withColumn(
        "data",
        to_date(col("data"), "dd/MM/yyyy")
    )
    .withColumn(
        "valor",
        col("valor").cast("double")
    )
    .dropDuplicates(["data"])
    .filter(
        col("data").isNotNull()
        & col("valor").isNotNull()
    )
)

display(df_dolar_silver.limit(10))
```

```python

print("Registros da Selic:")
print(df_selic_silver.count())

print("Registros do dólar:")
print(df_dolar_silver.count())

print("Valores nulos na Selic:")
display(
    df_selic_silver.filter(
        col("data").isNull() | col("valor").isNull()
    )
)

print("Valores nulos no dólar:")
display(
    df_dolar_silver.filter(
        col("data").isNull() | col("valor").isNull()
    )
)
```

```python

(
    df_selic_silver.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable("workspace.economic_data.silver_selic")
)

(
    df_dolar_silver.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable("workspace.economic_data.silver_dolar")
)

print("Tabelas Silver criadas com sucesso!")
```
